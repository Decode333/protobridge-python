# Chapter 4 - Filtering, Selection & Sorting

```python
euro2012 = pd.read_csv("Euro_2012_stats_TEAM.csv")
euro2012
```

```python
# filtering
euro2012_filtered = euro2012[["Team","Goals","Red Cards","Yellow Cards"]]
euro2012_filtered
```

```python
# Get top 5 teams received yellow cards
euro2012_filtered.sort_values(by="Yellow Cards",ascending=False)
```

```python
# Get top 5 teams received red cards
euro2012_filtered.sort_values(by="Red Cards",ascending=False)
```

```python
# Get teams scored only two goals
euro2012_filtered[euro2012_filtered["Goals"] == 2]
```

```python
# Get teams with more than 5 yellow cards and sort by max goals
(euro2012_filtered[["Team","Yellow Cards","Goals"]][euro2012_filtered["Yellow Cards"] > 5]).sort_values("Goals",ascending=False)
```

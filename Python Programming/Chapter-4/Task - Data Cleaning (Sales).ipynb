# Import required libraries

```python
import pandas as pd
import numpy as np
from datetime import datetime
```

# Load the datasets

```python
salesData_2017 = pd.read_csv(r"Sales Data/Sales Transactions-2017.csv")
salesData_2018 = pd.read_csv(r"Sales Data/Sales Transactions-2018.csv")
salesData_2019 = pd.read_csv(r"Sales Data/Sales Transactions-2019.csv")
```

```python
# combine all the data into one
bronzeSalesData = pd.concat([salesData_2017,salesData_2018,salesData_2019])
bronzeSalesData.head()
```

# Explore data for initial analysis

```python
print(f"Dimensions: {bronzeSalesData.shape}")
print(f"Column Datatypes: \n{bronzeSalesData.dtypes}")
print(f"Columns:\t{bronzeSalesData.columns}")
```

```python
print(f"Null data info:\n{bronzeSalesData.isnull().sum()}")
print(f"Statistical analysis on bronze data\n")
bronzeSalesData.describe(include='all')
```

#### Observations
The datatypes of the columns need to cleared for proper statistical information
- Date - to be updated as datetime
- Qty - to be updated as integer
- Rate, Gross, Disc, Voucher Amount - to be updated as float
- Disc - null can be updated as 0s

### Copy the raw data for initial cleaning

```python
silverSalesData = bronzeSalesData.copy()
```

```python
silverSalesData
```

```python
silverSalesData = silverSalesData.replace(r"^\s*$", np.nan, regex=True) # to convert the empty string as proper NaN values
# convert all the numeric columns into numeric from string or object datatype
numeric_cols = [
    "Qty",
    "Rate",
    "Gross",
    "Disc",
    "Voucher Amount"
]

for col in numeric_cols:
    silverSalesData[col] = (
        silverSalesData[col]
        .astype("string")
        .str.replace(",", "", regex=False)
        .str.strip()
    )

    silverSalesData[col] = pd.to_numeric(
        silverSalesData[col],
        errors="coerce"
    )
```

```python
# convert date into datetime datatype
silverSalesData["Date"] = pd.to_datetime(
    bronzeSalesData["Date"]
        .astype("string")
        .str.strip(),
    format="%d/%m/%Y",
    errors="coerce"
)
silverSalesData.dtypes
```

```python
silverSalesData.head()
```

```python
for col in numeric_cols:
    source_not_empty = bronzeSalesData[col].notna()
    conversion_failed = source_not_empty & silverSalesData[col].isna()

    print(
        f"{col}: {conversion_failed.sum()} "
        "non-empty values failed numeric conversion"
    )
```

```python
def normalize_source(series):
    return (
        series
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )
```

```python
conversion_failure_masks = {}

for col in numeric_cols:
    source_normalized = normalize_source(bronzeSalesData[col])

    failure_mask = (
        source_normalized.notna()
        & silverSalesData[col].isna()
    )

    conversion_failure_masks[col] = failure_mask

    print(
        f"{col}: {failure_mask.sum()} genuine conversion failures"
    )
```

```python
print("Minimum date:", silverSalesData["Date"].min())
print("Maximum date:", silverSalesData["Date"].max())
print("Invalid dates:", silverSalesData["Date"].isna().sum())
```

```python
missingSummary = pd.DataFrame({
    "Missing Count": silverSalesData.isna().sum(),
    "Missing Percentage": (
        silverSalesData.isna().mean() * 100
    ).round(2)
})

missingSummary.sort_values(
    "Missing Percentage",
    ascending=False
)
```

```python
# Fill missing discount with Zero
silverSalesData["Disc"] = (
    silverSalesData["Disc"].fillna(0)
)
print(
    "Missing discount after filling:",
    silverSalesData["Disc"].isna().sum()
)
```

```python
print(silverSalesData["Disc"].describe())
silverSalesData["Disc"].value_counts().head(10)
```

```python
critical_transaction_cols = [
    "Date",
    "Voucher",
    "Party",
    "Product",
    "Qty",
    "Rate",
    "Gross"
]

transaction_mask = (
    silverSalesData[critical_transaction_cols]
    .notna()
    .all(axis=1)
)
```

```python
print("Total rows:", len(silverSalesData))
print("Valid transaction rows:", transaction_mask.sum())
print("Other rows:", (~transaction_mask).sum())
```

```python
rejectedSalesData = (
    silverSalesData.loc[~transaction_mask]
    .copy()
)

validSalesData = (
    silverSalesData.loc[transaction_mask]
    .copy()
)
rejectedSalesData[
    [
        "Date",
        "Voucher",
        "Party",
        "Product",
        "Qty",
        "Rate",
        "Gross",
        "Disc",
        "Voucher Amount"
    ]
].head(50)
```

```python
rejectedSalesData.isna().sum()
```

```python
def get_missing_columns(row):
    return ", ".join(
        row.index[row.isna()].tolist()
    )

rejectedSalesData["Missing Columns"] = (
    rejectedSalesData.apply(
        get_missing_columns,
        axis=1
    )
)

rejectedSalesData[
    "Missing Columns"
].value_counts().head(20)
```

## Distinguish comment rows from invalid transactions
A comment or continuation row typically has no date, voucher, product, quantity, rate, or gross.

```python
comment_or_continuation_mask = (
    silverSalesData["Date"].isna()
    & silverSalesData["Voucher"].isna()
    & silverSalesData["Product"].isna()
    & silverSalesData["Qty"].isna()
    & silverSalesData["Rate"].isna()
    & silverSalesData["Gross"].isna()
)
```

A potentially invalid transaction has identifying information but is missing a critical numeric value:

```python
invalid_transaction_mask = (
    silverSalesData["Date"].notna()
    & silverSalesData["Voucher"].notna()
    & silverSalesData["Product"].notna()
    & (
        silverSalesData["Qty"].isna()
        | silverSalesData["Rate"].isna()
        | silverSalesData["Gross"].isna()
    )
)
```

```python
print(
    "Comments or continuation rows:",
    comment_or_continuation_mask.sum()
)

print(
    "Potential invalid transactions:",
    invalid_transaction_mask.sum()
)
```

```python
silverSalesData.loc[
    invalid_transaction_mask,
    [
        "Date",
        "Voucher",
        "Party",
        "Product",
        "Qty",
        "Rate",
        "Gross",
        "Disc",
        "Voucher Amount"
    ]
].head(50)
```

```python
silverSalesData.loc[
    comment_or_continuation_mask,
    [
        "Date",
        "Voucher",
        "Party",
        "Product",
        "Qty",
        "Rate",
        "Gross",
        "Disc",
        "Voucher Amount"
    ]
].head(50)
```

### Create a row-status column
This provides an auditable cleaning process.

```python
silverSalesData["Row Status"] = "Needs Review"

silverSalesData.loc[
    transaction_mask,
    "Row Status"
] = "Valid Transaction"

silverSalesData.loc[
    comment_or_continuation_mask,
    "Row Status"
] = "Comment or Continuation"

silverSalesData.loc[
    invalid_transaction_mask,
    "Row Status"
] = "Invalid Numeric Transaction"
```

```python
silverSalesData["Row Status"].value_counts()
```

Then split the data:

```python
rejectedSalesData = (
    silverSalesData[
        silverSalesData["Row Status"]
        != "Valid Transaction"
    ]
    .copy()
)

silverSalesData = (
    silverSalesData[
        silverSalesData["Row Status"]
        == "Valid Transaction"
    ]
    .copy()
)

silverSalesData.drop(
    columns=["Row Status"],
    inplace=True
)
```

```python
silverSalesData.reset_index(
    drop=True,
    inplace=True
)

rejectedSalesData.reset_index(
    drop=True,
    inplace=True
)
```

### Clean text columns
Now standardize whitespace in the text fields.

```python
text_cols = [
    "Voucher",
    "Party",
    "Product"
]

for col in text_cols:
    silverSalesData[col] = (
        silverSalesData[col]
        .astype("string")
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )
```

Preserve the original names while creating standardized versions for grouping:

Because these values might refer to the same party:
- Srinu
- SRINU
- srinu

```python
silverSalesData["Party Standardized"] = (
    silverSalesData["Party"]
    .str.upper()
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

silverSalesData["Product Standardized"] = (
    silverSalesData["Product"]
    .str.upper()
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)
```

## Analyze exact duplicate rows

```python
exact_duplicate_count = (
    silverSalesData.duplicated().sum()
)

print(
    "Exact duplicate rows:",
    exact_duplicate_count
)
```

```python
duplicateRows = (
    silverSalesData[
        silverSalesData.duplicated(
            keep=False
        )
    ]
    .sort_values(
        by=[
            "Date",
            "Voucher",
            "Party",
            "Product"
        ]
    )
)

duplicateRows.head(50)
```

```python
# drop the duplicates
silverSalesData = (
    silverSalesData
    .drop_duplicates()
    .reset_index(drop=True)
)
```

## Validate Gross = Qty × Rate

```python
silverSalesData["Calculated Gross"] = (
    silverSalesData["Qty"]
    * silverSalesData["Rate"]
)
silverSalesData["Gross Difference"] = (
    silverSalesData["Gross"]
    - silverSalesData["Calculated Gross"]
)
silverSalesData["Gross Valid"] = np.isclose(
    silverSalesData["Gross"],
    silverSalesData["Calculated Gross"],
    rtol=0,
    atol=0.01
)
```

```python
silverSalesData[
    "Gross Valid"
].value_counts()
```

```python
silverSalesData.loc[
    ~silverSalesData["Gross Valid"],
    [
        "Date",
        "Voucher",
        "Product",
        "Qty",
        "Rate",
        "Gross",
        "Calculated Gross",
        "Gross Difference"
    ]
].head(20)
```

## Create basic validity flags

```python
silverSalesData["Invalid Qty"] = (
    silverSalesData["Qty"] <= 0
)

silverSalesData["Invalid Rate"] = (
    silverSalesData["Rate"] < 0
)

silverSalesData["Invalid Gross"] = (
    silverSalesData["Gross"] < 0
)

silverSalesData["Invalid Discount"] = (
    silverSalesData["Disc"] < 0
)
```

```python
validationSummary = pd.Series({
    "Invalid Qty": silverSalesData[
        "Invalid Qty"
    ].sum(),

    "Invalid Rate": silverSalesData[
        "Invalid Rate"
    ].sum(),

    "Invalid Gross": silverSalesData[
        "Invalid Gross"
    ].sum(),

    "Invalid Discount": silverSalesData[
        "Invalid Discount"
    ].sum(),

    "Gross Mismatch": (
        ~silverSalesData["Gross Valid"]
    ).sum()
})

validationSummary
```

```python
# Missing discount means no recorded discount
silverSalesData["Disc"] = (
    silverSalesData["Disc"].fillna(0)
)

# Identify complete transaction rows
critical_transaction_cols = [
    "Date",
    "Voucher",
    "Party",
    "Product",
    "Qty",
    "Rate",
    "Gross"
]

transaction_mask = (
    silverSalesData[critical_transaction_cols]
    .notna()
    .all(axis=1)
)

print("Total rows:", len(silverSalesData))
print("Valid transaction rows:", transaction_mask.sum())
print("Rows requiring review:", (~transaction_mask).sum())
```

#### TASK 1
Give 2 examples of all String functions which should be meaningful.

```python
# From name, generate username
fname = "Nithin"
lname = "Raj"

username = ".".join([fname.lower(), lname.lower()])

print(username)
```

```python
# validate email address

email = "nithin.raj@example.com"
is_valid = email.count("@") == 1
domain = email.split("@")[1]
domainEndsWithCom = domain.endswith(".com")
is_valid = is_valid and domainEndsWithCom
print(f"Is email \"{email}\" valid? {is_valid}")
print(f"Domain: {domain}")
```

```python
# Hashtag generator
tagContent = "Justice for all"
hashtag = "#" + tagContent.title().replace(" ", "")
print(hashtag)
```

```python
# censor unwanted words
text = "This is a bad example of a bad situation."
unwanted_word = "bad"
censored_text = text.replace(unwanted_word, "*" * len(unwanted_word))
print(censored_text)
```

```python
# Search for keyword in a sentence
sentence = "The quick brown fox jumps over the lazy dog."
keyword1 = "fox"
keyword2 = "nithin"
print("Sentence: " + sentence)
for keyword in [keyword1,keyword2]:
    found = sentence.find(keyword)
    if found != -1:
        print(f"Keyword \"{keyword}\" found at index {found}.")
    else:
        print(f"Keyword \"{keyword}\" not found.")
```

```python
# mask data
cardNumber = "1234567898765432"
maskedCardNumber = "*" * (len(cardNumber)-4) + cardNumber[len(cardNumber)-4:]
print(f"Masked Card Number: {maskedCardNumber}")
```

```python
# password strength validation
password = "Nithi23"
passLen = len(password)

digit = False
lowercase = False
uppercase = False
for char in password:
    if passLen<8:
        print("Password is too short.")
        break
    if char.isalnum():
        if char.isdigit():
            digit = True
        elif char.islower():
            lowercase = True
        elif char.isupper():
            uppercase = True
if digit and lowercase and uppercase:
    print("Password is strong.")
else:
    print("Password is not strong.")
```

```python
# while loop
import random

price = random.randint(100,1000)
total = 0
while total < 1000:
    item_price = price
    total += item_price
print("Total checkout amount: "+str(total)+". Eligible for free delivery")
```

```python
# for loop, if-else, for loop with in operator, for loop with if statement,if with in operator
employeesEligible = ["John", "Sarah", "Bhuvan", "Tina"]
employees = ["John", "Nithin", "Tina", "Bhuvan","Sarah","Ramesh"]
for employee in employees:
    if employee in employeesEligible:
        print(f"{employee} is eligible for bonus.")
    else:
        print(f"{employee} is not eligible for bonus.")
```

```python
# if elif else
order_total = 650
if order_total >= 1000:
    print("Free Delivery")
elif order_total >= 500:
    print("Delivery Charge ₹50")
else:
    print("Delivery Charge ₹100")
```

user defined functions without for loop, if loop

```python
def generateUsername(first_name, last_name):
    return ".".join([first_name.lower(), last_name.lower()])

fname = "Nithin"
lname = "Raj"
print("Your username is: "+generateUsername(fname, lname))
```

user defined functions with for loop

```python
def calculateDiscount(price,discount):
    discountedPrice = price*(1-discount/100)
    return discountedPrice

checkoutPrice = 1000
discount = 10
finalPrice = calculateDiscount(checkoutPrice,discount)
print(f"Checkout Price: {checkoutPrice}, Discount: {discount}%, Final Price: {finalPrice}")
```

user defined functions with for loop and if condition

```python
def calculateCheckoutPrice(prices):
    total=0
    for price in prices:
        if price>500:
            total += price*0.2
        else:
            total += price
    return total

itemPrices = [10,56,450,410,1000]
checkoutPrice = calculateCheckoutPrice(itemPrices)
print(f"Total checkout price: {checkoutPrice}")
```

for loop with not in operator

```python
available = ["Laptop", "Mouse"]

for product in ["Laptop", "Keyboard", "Mouse"]:
    if product not in available:
        print(product, "is unavailable")
```

```python
# continue
days = ["Monday", "Tuesday", "Saturday", "Sunday"]

for day in days:
    if day in ["Saturday", "Sunday"]:
        continue
    print("Send notification on", day)
```

```python
# break
products = ["Mouse", "Keyboard", "Laptop", "Monitor"]

for product in products:
    if product == "Laptop":
        print("Product Found")
        break
```

# List Comprehension

#### Example 1

```python
numbers = [1, 2, 3, 4, 5, 6]

even_squares = [
    number ** 2
    for number in numbers
    if number % 2 == 0
]

print(even_squares)
```

```python
gross_sales = [
    3380.00,
    9720.00,
    11500.00,
    8450.00,
    1320.00
]

high_value_sales = [
    sale
    for sale in gross_sales
    if sale >= 5000
]

print(high_value_sales)
```

# Lambda function

```python
cube = lambda number: number ** 3

print(cube(5))
```

```python
increased_sales = list(
    map(
        lambda sale: round(sale * 1.05, 2),
        gross_sales
    )
)

print(increased_sales)
```

```python
high_value_sales = list(
    filter(
        lambda sale: sale >= 5000,
        gross_sales
    )
)

print(high_value_sales)
```

```python
high_value_sales = list(
    map(
        lambda sale: sale >= 5000,
        gross_sales
    )
)

print(high_value_sales)
```

The map only returned the boolean output on the condition. To get the actual values that are True for the condition can be produced by filter

```python
from functools import reduce
```

```python
total_gross_sales = reduce(
    lambda total, sale: total + sale,
    gross_sales,
    0
)

print(total_gross_sales)
```

# Merging Dataframes

The row merging demonstrated on the data cleaning task. Now showing merging by columns.

```python
import pandas as pd
```

```python
employees = {
    "emp_id":["ab12390","bd78238","hj75634","yu87349","po09347"],
    "emp_designation":["Senior Associate","Lead","Associate","Lead","Delivery Head"],
    "emp_age":[32,35,28,33,39],
    "emp_deptId":["HR","ET","ET","SM","ET"],
    "emp_salary":[45000,150000,100000,135000,295000]
}

dept = {
    "dept_id":["HR","ET","SM"],
    "dept_name":["Human Resources","Engineering & Technology","Sales Management"],
    "dept_reporting":["CFO","CTO","CEO"]
}
```

```python
employees_data = pd.DataFrame(data=employees)
employees_data
```

```python
dept_data = pd.DataFrame(data=dept)
dept_data
```

```python
data_merged = pd.merge(
    employees_data,
    dept_data,how="inner",
    left_on="emp_deptId",
    right_on="dept_id"
)
data_merged
```

# Python Data Structures

### Python Datatype - List

```python
# Add products to the cart

cart = [] # intialize empty cart (list)
cart.append('Apple') # add Apple to the cart
cart.append('Banana')

# print the items in the cart
print(f"Items in the cart: {cart}")

# clear the cart
print("Clearing the cart...")
cart.clear()
print(f"Items in the cart: {cart}")
```

```python
# Add new products to the inventory
inventory = ['Apple', 'Banana', 'Orange']
new_products = ['Mango', 'Kiwi']
print("Items before adding new products to the inventory: ", inventory)
inventory.extend(new_products)
print("Items after adding new products to the inventory: ", inventory)
```

```python
# insert an order in a specific position in the list
orders = ["Order1", "Order2", "Order3"]

orders.insert(1, "Order1.b")
print("Inserted order. Updated orders list: ",orders)
orders.insert(5, "Order5") #adding an order at the end of the list
print("Inserted order. Updated orders list: ",orders)
orders.insert(2, "Order6")
print("Inserted order. Updated orders list: ",orders)
print("Cancel an order")
orders.remove("Order2")
print("Updated Orders list: ",orders)
```

```python
# deliver the order and remove it from the list
print("Delivering order in queue: ",orders.pop())
print("Delivering priority order: ", orders.pop(2))
print("Pending orders: ", orders)
```

```python
orders.append("Order1")
# Duplicate check
seen = list()
for order in orders:

    if order not in seen:
        seen.append(order)
        if orders.count(order)>1:
            print(f"The order {order} is duplicated {orders.count(order)} times")
```

```python
student1_marks = [302,902,356,778,834,476,509]

# copy the marks of student 1 to student 2
student2_marks = student1_marks.copy()

# sort to find the first mark
student1_marks.sort()    # ascending
print("First Mark of student 1: ",student1_marks[len(student1_marks)-1])

student1_marks.sort(reverse=True)    # descending
print("First Mark of student 1: ",student1_marks[0])
print("Marks of Student 1: ",student1_marks)
print("Marks of Student 2: ",student2_marks)
```

```python
# remove a mark of student 2
student2_marks.remove(902)
print("Marks of Student 1: ",student1_marks)
print("Marks of Student 2: ",student2_marks)
```

```python
# remove a mark of student 1
student1_marks.remove(834)
print("Marks of Student 1: ",student1_marks)
print("Marks of Student 2: ",student2_marks)
```

Without touching the marks of one student, manipulate the other's

```python
student1_marks = [302,902,356,778,834,476,509]

# assign the marks of student 1 to student 2
student2_marks = student1_marks

# sort to find the first mark
student1_marks.sort()    # ascending
print("First Mark of student 1: ",student1_marks[len(student1_marks)-1])

student1_marks.sort(reverse=True)    # descending
print("First Mark of student 1: ",student1_marks[0])
print("Marks of Student 1: ",student1_marks)
print("Marks of Student 2: ",student2_marks)
```

### Python Datatype - Set

```python
print(orders)
```

```python
# remove duplicate orders
orders = set(orders)
print("Unique orders: ",orders)
```

```python
# unique visitors
block1_visitors = {"Senthil","Rajan"}
print("Existing block 1 visitors: ",block1_visitors)
print("New visitor to 1: Anbu")
block1_visitors.add("Anbu")
print("updated visitors in block 1: ",block1_visitors)
```

```python
block2_visitors = {"Velu","Guna","Anbu"}
print("Common member to block 1 & 2: ", block2_visitors.intersection(block1_visitors))
```

```python
print("All visitors to both the blocks: ",block1_visitors.union(block2_visitors))
```

```python
head_block = {"Rajan","Chandra"}
print("Main head block: ",head_block)

# unsubscribe from one block and add in another
user = head_block.difference(block1_visitors)
print(user," unsubscribed to join block 2")
block2_visitors.update(user)
print("updated visitors in block 1: ",block2_visitors)
```

### Python Datatype - Dictionary

```python
# company employee records
employee1 = {
    "name": "John",
    "salary": 50000,
    "role": "Admin"
}

print(employee1['name'])
```

```python
# Increment to employee1
employee1['salary'] = 100000
print(f"Salary of {employee1['name']} incremented as {employee1['salary']}")
```

```python
employee2 = {
    "name": "Peter",
    "salary": 60000,
    "role": "Dev"
}
print("Details of employee2: ",employee2.values())
```

```python
# add info to employees

employee2["location"] = "Chennai"
employee1["location"] = "Pune"

print(f"Employees updated info: \n{employee1}\n{employee2}")
```

```python
employee1.setdefault("type","probation")
employee2.setdefault("type","permanent")

print(f"Employees current status:\n"
      f"{employee1['name']} is {employee1['type']}\n"
      f"{employee2['name']} is {employee2['type']}"
      )

# promoting employee1
employee1["type"] = "permanent"
print(f"Employees current status:\n"
      f"{employee1['name']} is {employee1['type']}\n"
      f"{employee2['name']} is {employee2['type']}"
      )
```

```python
employee3 = {'name': 'Emily', 'salary': 103000, 'role': 'Hospitality', 'location': 'Kochi','type':'permanent'}
print(f"Employees current status:\n"
      f"{employee1['name']} is {employee1['type']}\n"
      f"{employee2['name']} is {employee2['type']}\n"
      f"{employee3['name']} is {employee3['type']}"
      )
# employee2 resigned
employee2.pop("type")

def checkEmployeeStatus(emp):
    if "type" in emp.keys():
        print(f"The employee '{emp['name']}' is working.")
    else:
        print(f"The employee '{emp['name']}' has resigned.")

checkEmployeeStatus(employee1)
checkEmployeeStatus(employee2)
checkEmployeeStatus(employee3)
```

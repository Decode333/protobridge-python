```python
silverSalesData = silverSalesData.replace(r"^\s*$", np.nan, regex=True) # to convert the empty string as proper NaN values
# convert all the numeric columns into numeric from string or object datatype
numeric_cols = [
    "Qty",
    "Rate",
    "Gross",
    "Disc",
    "Voucher Amount"
]

for col in numeric_cols:
    silverSalesData[col] = (
        silverSalesData[col]
        .astype("string")
        .str.replace(",", "", regex=False)
        .str.strip()
    )

    silverSalesData[col] = pd.to_numeric(
        silverSalesData[col],
        errors="coerce"
    )
```

```python
# convert date into datetime datatype
silverSalesData["Date"] = pd.to_datetime(
    bronzeSalesData["Date"]
        .astype("string")
        .str.strip(),
    format="%d/%m/%Y",
    errors="coerce"
)
silverSalesData.dtypes
```

```python
silverSalesData.head()
```

```python
for col in numeric_cols:
    source_not_empty = bronzeSalesData[col].notna()
    conversion_failed = source_not_empty & silverSalesData[col].isna()

    print(
        f"{col}: {conversion_failed.sum()} "
        "non-empty values failed numeric conversion"
    )
```

```python
def normalize_source(series):
    return (
        series
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )
```

```python
conversion_failure_masks = {}

for col in numeric_cols:
    source_normalized = normalize_source(bronzeSalesData[col])

    failure_mask = (
        source_normalized.notna()
        & silverSalesData[col].isna()
    )

    conversion_failure_masks[col] = failure_mask

    print(
        f"{col}: {failure_mask.sum()} genuine conversion failures"
    )
```

#### Voucher summary

```python
voucherSummary = (
    silverSalesData
    .groupby(
        [
            "Transaction Key",
            "Date",
            "Voucher",
            "Party Standardized"
        ],
        as_index=False
    )
    .agg(
        Line_Count=("Product", "size"),
        Product_Count=("Product Standardized", "nunique"),
        Quantity_Total=("Qty", "sum"),
        Gross_Total=("Gross", "sum"),
        Discount_Total=("Disc", "sum"),

        # Count how many rows contain a reported voucher amount
        Voucher_Amount_Count=("Voucher Amount", "count"),

        # Get the minimum and maximum to detect differing totals
        Voucher_Amount_Min=("Voucher Amount", "min"),
        Voucher_Amount_Max=("Voucher Amount", "max")
    )
)
```

#### Derive the reported voucher amount
- If one voucher total is recorded on only one transaction line, minimum and maximum will be the same.

```python
voucherSummary["Reported Voucher Amount"] = (
    voucherSummary["Voucher_Amount_Max"]
)
```

```python
voucherSummary["Calculated Net Amount"] = (
    voucherSummary["Gross_Total"]
    - voucherSummary["Discount_Total"]
)
```

```python
voucherSummary["Voucher Difference"] = (
    voucherSummary["Reported Voucher Amount"]
    - voucherSummary["Calculated Net Amount"]
)

voucherSummary["Voucher Difference Abs"] = (
    voucherSummary["Voucher Difference"].abs()
)
```

#### Identify missing or inconsistent voucher totals

```python
voucherSummary["Voucher Amount Missing"] = (
    voucherSummary["Reported Voucher Amount"].isna()
)
voucherSummary["Multiple Voucher Amount Entries"] = (
    voucherSummary["Voucher_Amount_Count"] > 1
)
```

```python
voucherSummary["Conflicting Voucher Amounts"] = (
    voucherSummary["Voucher_Amount_Count"].gt(1)
    & ~np.isclose(
        voucherSummary["Voucher_Amount_Min"],
        voucherSummary["Voucher_Amount_Max"],
        rtol=0,
        atol=0.01,
        equal_nan=True
    )
)
```

#### Validate the voucher amount

```python
voucherSummary["Voucher Valid"] = np.isclose(
    voucherSummary["Reported Voucher Amount"],
    voucherSummary["Calculated Net Amount"],
    rtol=0,
    atol=1.00,
    equal_nan=False
)
```

#### Classisifcation of the reconcilation results

```python
voucherSummary["Voucher Reconciliation Status"] = np.select(
    [
        voucherSummary["Voucher Amount Missing"]
        .fillna(False)
        .to_numpy(dtype=bool),

        voucherSummary["Conflicting Voucher Amounts"]
        .fillna(False)
        .to_numpy(dtype=bool),

        voucherSummary["Voucher Valid"]
        .fillna(False)
        .to_numpy(dtype=bool),

        voucherSummary["Voucher Difference Abs"]
        .le(5.00)
        .fillna(False)
        .to_numpy(dtype=bool),

        voucherSummary["Voucher Difference Abs"]
        .le(20.00)
        .fillna(False)
        .to_numpy(dtype=bool)
    ],
    [
        "Missing Reported Voucher Amount",
        "Conflicting Reported Voucher Amounts",
        "Reconciled",
        "Small Difference",
        "Moderate Difference"
    ],
    default="Material Difference"
)
```

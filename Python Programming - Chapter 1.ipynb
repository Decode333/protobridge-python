```python
# From name, generate username
fname = "Nithin"
lname = "Raj"

username = ".".join([fname.lower(), lname.lower()])

print(username)
```

```python
# validate email address

email = "nithin.raj@example.com"
is_valid = email.count("@") == 1
domain = email.split("@")[1]
domainEndsWithCom = domain.endswith(".com")
is_valid = is_valid and domainEndsWithCom
print(f"Is email \"{email}\" valid? {is_valid}")
print(f"Domain: {domain}")
```

```python
# Hashtag generator
tagContent = "Justice for all"
hashtag = "#" + tagContent.title().replace(" ", "")
print(hashtag)
```

```python
# censor unwanted words
text = "This is a bad example of a bad situation."
unwanted_word = "bad"
censored_text = text.replace(unwanted_word, "*" * len(unwanted_word))
print(censored_text)
```

```python
# Search for keyword in a sentence
sentence = "The quick brown fox jumps over the lazy dog."
keyword1 = "fox"
keyword2 = "nithin"
print("Sentence: " + sentence)
for keyword in [keyword1,keyword2]:
    found = sentence.find(keyword)
    if found != -1:
        print(f"Keyword \"{keyword}\" found at index {found}.")
    else:
        print(f"Keyword \"{keyword}\" not found.")
```

```python
# mask data
cardNumber = "1234567898765432"
maskedCardNumber = "*" * (len(cardNumber)-4) + cardNumber[len(cardNumber)-4:]
print(f"Masked Card Number: {maskedCardNumber}")
```

```python
# password strength validation
password = "Nithi23"
passLen = len(password)

digit = False
lowercase = False
uppercase = False
for char in password:
    if passLen<8:
        print("Password is too short.")
        break
    if char.isalnum():
        if char.isdigit():
            digit = True
        elif char.islower():
            lowercase = True
        elif char.isupper():
            uppercase = True
if digit and lowercase and uppercase:
    print("Password is strong.")
else:
    print("Password is not strong.")
```
